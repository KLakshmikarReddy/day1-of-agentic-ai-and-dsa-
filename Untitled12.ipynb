```python

def prepare_input(text):
    """Clean user input."""
    return text.strip()


def generate_response(text):
    """Apply simple rules to the input."""
    if not text:
        return "Please enter a question."

    question = text.lower()

    if "machine learning" in question:
        return (
            "Machine learning is a branch of AI "
            "that learns patterns from data."
        )

    if "llm" in question:
        return (
            "LLM stands for Large Language Model. "
            "It generates text using learned patterns."
        )
    if "rag" in question:
     return (
        "RAG stands for Retrieval-Augmented Generation. "
        "It combines information retrieval with language generation."
    )
    if "agent" in question:
     return (
        "An AI agent uses an AI model and tools "
        "to work toward a goal."
    )

    return f"You asked: {text}"


user_input = input("Ask a question: ")
clean_input = prepare_input(user_input)
answer = generate_response(clean_input)

print("\nAssistant:", answer)
```

```python
def reverse_string(s):
  chars=list(s)
  left=0
  right=len(chars)-1
  while left<right:
    chars[left],chars[right]=chars[right],chars[left]
    left+=1
    right-=1
  return "".join(chars)

print(reverse_string("hello"))
```

```python
def is_palindrome(s):
    left = 0
    right = len(s) - 1

    while left < right:
        if s[left] != s[right]:
            return False

        left += 1
        right -= 1

    return True

print(is_palindrome("madam"))
```
